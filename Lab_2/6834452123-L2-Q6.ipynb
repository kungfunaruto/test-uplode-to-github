## Q6
จงเขียนโปรแกรมที่รับน้ำหนัก และส่วนสูง จากผู้ใช้ มาคำนวณ bmi โดยให้ระบุหน่วยของน้ำหนักเป็น kg
หรือ lbs และ หน่วยของส่วนสูงเป็น cm หรือ m หรือ ft แล้วไปแปลงเป็น kg และ m และคำนวณแสดงผลลัพธ์ว่าผู้ใช้มีรูปร่างในระดับใด <br>
(1 kg. = 2.205 lbs, 1 m. = 3.2808399 feet)

### ตัวอย่างการทำงาน: 

Weight : **61.3 kg**<br>
Height : **1.65 m**<br>
รูปร่างปกติ

### ตัวอย่างการทำงาน: 

Weight : **159.3 lbs** <br>
Height : **4.92 ft**<br>
อ้วนระดับ 2

### ตัวอย่างการทำงาน:

Weight : **110.8 lbs** <br> Height : **167 cm**<br>ผอม


```python
def run():
    weight, wunit= input('Weight (kg.) : ').split(' ')
    height, hunit = input('Height (m.) : ').split(' ')
    weight = float(weight)
    height = float(height)

    #convert unit weight
    if wunit == 'kg' :
        weight = weight
    elif wunit == 'lbs' : 
        weight = weight/2.205
    else : 
        print('error')
    
    #convert unit height
    if hunit == 'm' : 
        height = height
    elif hunit == 'cm' :
        height = height/100
    elif hunit == 'ft':
        height = height/3.2808399
    else : 
        print('error')

    bmi = weight/(height**2)
    #print(bmi)
    if bmi < 18.5:
        print('ผอม')
    elif bmi < 23:
        print('รูปร่างปกติ')
    elif bmi < 25:
        print('รูปร่างอ้วน')
    elif bmi < 30:
        print('อ้วนระดับ 1')
    else:
        print('อ้วนระดับ 2')
```

```python
'''ตรวจสอบผลลัพธ์ของโปรแกรมด้วยการรันที่ cell นี้'''
run()
```

```python
'''ตรวจสอบผลการคำนวณ'''
from unittest.mock import patch
from io import StringIO

def test(inputs,expected_output): 
    with patch('builtins.input', side_effect=inputs), patch('sys.stdout', new_callable=StringIO) as mock_stdout:
        run()
        assert mock_stdout.getvalue() == expected_output,mock_stdout.getvalue()
def test_easy_print(inputs,expected_output):
    with patch('builtins.input', side_effect=inputs), patch('sys.stdout', new_callable=StringIO) as mock_stdout:
        run()
        assert expected_output in mock_stdout.getvalue(), mock_stdout.getvalue()
```

```python
'''ตรวจสอบการแสดงผลลัพธ์อย่างง่าย'''
test_easy_print(inputs=['61.3 kg','1.65 m'],expected_output='รูปร่าง')
print('ok')
```

```python
'''ตรวจสอบผลการคำนวณและการแสดงผลลัพธ์ให้เหมือนกับโจทย์ที่กำหนดให้'''
test(['61.3 kg','1.65 m'],'รูปร่างปกติ\n')  # 
test(['61.3 kg','165 cm'],'รูปร่างปกติ\n') 
test(['49.5 kg','5 ft'],'รูปร่างปกติ\n') 
test(['110.8 lbs','150 cm'],'รูปร่างปกติ\n') 
test(['110.8 lbs','1.50 m'],'รูปร่างปกติ\n') 
test(['92.5 lbs','4.8 ft'],'รูปร่างปกติ\n')
print('ok')
```
