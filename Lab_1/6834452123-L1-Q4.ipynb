## Q4
จงเขียนโปรแกรมที่รับชื่อและอายุจากผู้ใช้ แล้วคำนวณปีเกิดของผู้ใช้เพื่อสร้าง email address ของผู้ใช้โดย
เอาชื่อมาต่อด้วยเลข 2 หลักหลังของปีเกิดและต่อด้วย @chula.ac.th แสดงผลลัพธ์ดังตัวอย่างข้างล่างนี้

### ตัวอย่างการทำงาน: 

Name: Somsak <br>
Age : 20 <br>
Somsak48@chula.ac.th

### ตัวอย่างการทำงาน: 
        
Name: Jaidee <br>
Age : 65 <br>
Jaidee03@chula.ac.th

### ตัวอย่างการทำงาน: 
        
Name: Yaya <br>
Age : 71 <br>
Yaya97@chula.ac.th

```python
def run():
    from datetime import date
    name = input('Name: ')
    age = int(input('age: '))
    current_date = date.today()
    current_year_AD = current_date.year
    current_year_BE = current_year_AD + 543
    birth_year_BE = current_year_BE - age
    birth_year_BE2 =birth_year_BE % 100
    print(f'{name}{birth_year_BE2:02d}@chula.ac.th')
    
```

```python
'''ตรวจสอบผลลัพธ์ของโปรแกรมด้วยการรันที่ cell นี้'''
run()
```

```python
from unittest.mock import patch
from io import StringIO
from datetime import date 

def test(inputs,expected_output): 
    with patch('builtins.input', side_effect=inputs), patch('sys.stdout', new_callable=StringIO) as mock_stdout:
        run()
        assert mock_stdout.getvalue() == expected_output,mock_stdout.getvalue()
def test_easy_print(inputs,expected_output):
    with patch('builtins.input', side_effect=inputs), patch('sys.stdout', new_callable=StringIO) as mock_stdout:
        run()
        assert expected_output in mock_stdout.getvalue(), mock_stdout.getvalue()
```

```python
'''ตรวจสอบการแสดงผลลัพธ์อย่างง่าย'''
test_easy_print(inputs=['Somsak','65'],expected_output='Somsak')
print('ok')
```

```python
'''ตรวจสอบเลขสองตัวท้ายและการแสดงผลลัพธ์'''
'''กรณีเลขสองตัวท้ายของปีเกิดอยู่ในช่วง 00-09'''
test(['Somsak','68'],'Somsak00@chula.ac.th\n')
test(['Yaya','59'],'Yaya09@chula.ac.th\n') 
print('ok')
```

```python
'''ตรวจสอบเลขสองตัวท้ายและการแสดงผลลัพธ์'''
'''กรณีเลขสองตัวท้ายของปีเกิดอยู่ในช่วง 10-99'''
test(['Somsak','21'],'Somsak47@chula.ac.th\n')
test(['Jaidee','66'],'Jaidee02@chula.ac.th\n')  
test(['Yaya','72'],'Yaya96@chula.ac.th\n') 
print('ok')
```
