## Q3
จงเขียนฟังก์ชัน f(x) ที่คำนวณ 1/x! และฟังก์ชัน summation(a,b) ที่ใช้ฟังก์ชัน f และ คำนวณผลรวมของ f(x) เมื่อ x = a, a+1,..., b (นั่นคือ 1/a! + 1/(a+1)! + ... + 1/b!) โดยที่ a ต้องไม่มากกว่า b <br>
โดยใช้ ฟังก์ชัน factorial จาก module math <br>
 
จากนั้นให้เขียนโปรแกรมหลักที่รับ**จำนวนเต็มบวก 2 จำนวน m และ n** (m อาจมากกว่าหรือน้อยกว่า n) แล้วเรียกใช้ฟังก์ชัน summation เพื่อ คำนวณหาผลรวมของ f(x) เมื่อ x มีค่าตั้งแต่ m ถึง n <br>
**แสดงผลลัพธ์เป็นตัวเลขทศนิยม 4 ตำแหน่ง โดยใช้ ฟังก์ชัน round**

### ตัวอย่างการทำงาน: 

Enter m: **1** <br>
Enter n: **3** <br>
1.6667


### ตัวอย่างการทำงาน:

```python
# def f(x):

#     from math import factorial
#     P = 1/factorial(x)
#     return P

# def summation(a, b):
#     total = 0.0
#     for i in range(a, b + 1):
#         total += f(i)
#     return total

# def run():
#     m = int(input('Enter m: '))
#     n = int(input('Enter n: '))
#     if m < n :
#         final = summation(m, n)
#         print(round(final, 4))
#     elif m > n:
#         final = summation(n, m)
#         print(round(final, 4))


import math

def f(x):
    return 1 / math.factorial(x)

def summation(a, b):
    total = 0.00
    for i in range(a, b + 1):
        total += f(i)
    return total

def run():
    m = int(input("Enter m: "))
    n = int(input("Enter n: "))
    a, b = min(m, n), max(m, n)
    result = summation(a, b)
    print(round(result, 4))
```

```python
'''ตรวจสอบผลลัพธ์ของโปรแกรมด้วยการรันที่ cell นี้'''
run()
```

```python
'''ตรวจสอบผลการคำนวณ'''
from unittest.mock import patch
from io import StringIO
import math

def test(inputs,expected_output): 
    with patch('builtins.input', side_effect=inputs), patch('sys.stdout', new_callable=StringIO) as mock_stdout:
        run()
        assert mock_stdout.getvalue() == expected_output,mock_stdout.getvalue()
def test_easy_print(inputs,expected_output):
    with patch('builtins.input', side_effect=inputs), patch('sys.stdout', new_callable=StringIO) as mock_stdout:
        run()
        assert expected_output in mock_stdout.getvalue(), mock_stdout.getvalue()
```

```python
'''ตรวจสอบการแสดงผลลัพธ์อย่างง่าย'''
test_easy_print(inputs=['1','3'],expected_output='1')
print('ok')
```

```python
'''ตรวจสอบผลการคำนวณและการแสดงผลลัพธ์ให้เหมือนกับโจทย์ที่กำหนดให้'''
test(['1','3'],'1.6667\n')  
test(['3','6'],'0.2181\n') 
test(['5','6'],'0.0097\n')
test(['3','4'],'0.2083\n')
print('ok')
```
