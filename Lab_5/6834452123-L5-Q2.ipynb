จงเขียนโปรแกรมที่อ่านเฉลยของข้อสอบปรนัยจากไฟล์ sol\<n\>.txt (ข้อสอบมี 10 ข้อ ) 
และอ่านคำตอบของนักเรียนจากไฟล์ exam\<n\>.txt ที่มีคำตอบจากนักเรียนแต่ละคนอยู่ใน 1 บรรทัด
(ถือว่านักเรียนต้องตอบข้อสอบทุกข้อ)
และมีนักเรียนอยู่ 8 คน จากนั้นให้ตรวจคำตอบของนักเรียนทุกคน แล้วหา
- คะแนนของนักเรียนแต่ละคน
- คะแนนเฉลี่ยของข้อสอบแต่ละข้อ
- ข้อสอบข้อที่นักเรียนทำคะแนนได้น้อยสุด (แสดงให้ครบทุกข้อ หากคะแนนเท่ากันหลายข้อ )
- ข้อสอบข้อที่นักเรียนทำคะแนนได้มากสุด (แสดงให้ครบทุกข้อ หากคะแนนเท่ากันหลายข้อ )

**หมายเหตุ ให้ใช้ลิสต์ในข้อนี้ และสามารถใช้ฟังก์ชัน min และ max เพื่อหาค่าน้อยสุดและมากสุดในลิสต์ได้**

### ตัวอย่างการทำงาน: 
Choose your solution file: **sol1.txt** <br>
Choose your exam file: **exam1.txt** <br>
Student score : [9, 6, 9, 8, 7, 7, 9, 8]<br>

Question 1 : 0.625 <br>
Question 2 : 1.0 <br>
Question 3 : 0.5 <br>

Hardest : 3 <br>
Easiest : 2 6 <br>

```python
def run():
    sol_file = input("Choose your solution file: ")
    exam_file = input("Choose your exam file: ")

    # อ่านไฟล์เฉลย
    with open(sol_file, "r", encoding="utf-8") as f:
        solution = f.readline().strip().split()

    # อ่านไฟล์คำตอบนักเรียน
    with open(exam_file, "r", encoding="utf-8") as f:
        exams = [line.strip().split() for line in f]

    # ตรวจคำตอบนักเรียน
    student_scores = []
    for ans in exams:
        score = sum(1 for a, b in zip(solution, ans) if a == b)
        student_scores.append(score)

    print("Student score :", student_scores)
    print()  # เว้นบรรทัดตามโจทย์

    # คำนวณค่าเฉลี่ยรายข้อ
    n_students = len(exams)
    n_questions = len(solution)
    avg = []
    for q in range(n_questions):
        correct = sum(1 for ans in exams if ans[q] == solution[q])
        avg.append(correct / n_students)

    for i, a in enumerate(avg, start=1):
        print(f"Question {i} : {str(round(a,3))}")

    print()  # เว้นบรรทัดตามโจทย์

    # หาข้อที่ทำได้น้อยสุดและมากสุด
    min_val = min(avg)
    max_val = max(avg)
    hardest = [str(i+1) for i, a in enumerate(avg) if a == min_val]
    easiest = [str(i+1) for i, a in enumerate(avg) if a == max_val]

    print("Hardest :", " ".join(hardest))
    print("Easiest :", " ".join(easiest))
```

```python
'''ตรวจสอบผลลัพธ์ของโปรแกรมด้วยการรันที่ cell นี้'''
run()
```

```python
'''ตรวจสอบผลการคำนวณ'''
from unittest.mock import patch
from io import StringIO

def test(inputs,expected_output): 
    with patch('builtins.input', side_effect=inputs), patch('sys.stdout', new_callable=StringIO) as mock_stdout:
        run()
        assert mock_stdout.getvalue() == expected_output,mock_stdout.getvalue()
def test_easy_print(inputs,expected_output):
    with patch('builtins.input', side_effect=inputs), patch('sys.stdout', new_callable=StringIO) as mock_stdout:
        run()
        assert expected_output in mock_stdout.getvalue(), mock_stdout.getvalue()
```
